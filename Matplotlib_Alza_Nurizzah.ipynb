```python
# PIE CHART KOMPOSISI SUMBER PENDAPATAN

komponen = ['DBH_PAJAK','DBH_SDA','DAU_Grant', 'DAK_Fisik','PAD']

total = df[komponen].sum()

plt.figure(figsize=(8, 8))

plt.pie(total, labels=total.index, autopct='%1.1f%%',startangle=90)

plt.title('Komposisi Total Pendapatan Dari Dataset Fiskal',fontsize=15, fontweight='bold')

plt.show()
```

```python
# LINE CHART TREN IPM PER TAHUN

tren_ipm = df.groupby('Tahun')['IPM'].mean()

plt.figure(figsize=(10, 5))

plt.plot(tren_ipm.index, tren_ipm.values, marker='o', linewidth=2)

plt.title('Tren Rata-rata IPM per Tahun', fontsize=15, fontweight='bold')

plt.xlabel('Tahun')
plt.ylabel('Rata-rata IPM')

plt.xticks(tren_ipm.index, rotation=45)

plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# LINE CHART
# KEMISKINAN VS PENGANGGURAN

tren = df.groupby('Tahun')[['KEMISKINAN','PENGANGGURAN']].mean()

plt.figure(figsize=(10, 5))

plt.plot(
    tren.index,
    tren['KEMISKINAN'],
    marker='o',
    label='Kemiskinan')

plt.plot(
    tren.index,
    tren['PENGANGGURAN'],
    marker='s',
    linestyle='--',
    label='Pengangguran')

plt.title(
    'Tren Kemiskinan dan Pengangguran',
    fontsize=15,
    fontweight='bold')

plt.xlabel('Tahun')
plt.ylabel('Persentase')

plt.legend()

plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# HEATMAP KORELASI

kolom_korelasi = ['PAD','KEMISKINAN','PENGANGGURAN','KETIMPANGAN','IPM','bm_pendidikan','bm_kesehatan']

corr = df[kolom_korelasi].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)

plt.title('Heatmap Korelasi Variabel Fiskal dan Sosial', fontsize=15, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
#  MELIHAT 10 Daerah dengan PAD Terbesar Tahun 2022
# Membersihkan nama kolom
df.columns = df.columns.str.strip()

# Mengambil data tahun 2022
data_2022 = df[df['Tahun'] == 2022]

# Mengambil 10 daerah dengan PAD terbesar
data_top10 = data_2022.nlargest(10, 'PAD')

# Mengurutkan dari PAD terkecil ke terbesar
data_top10 = data_top10.sort_values('PAD')

# Data untuk grafik
kota = data_top10['Kota']
PAD = data_top10['PAD']

# Ukuran grafik
plt.figure(figsize=(8, 5))

# Membuat bar horizontal
plt.barh(kota, PAD, color='black')

# Judul dan label
plt.title('10 Daerah dengan PAD Terbesar Tahun 2022')
plt.xlabel('Pendapatan Asli Daerah (PAD)')
plt.ylabel('Daerah')

#Menambahkan nilai PAD di ujung setiap batang
for i, nilai in enumerate(PAD):
    plt.text(nilai + 5, i, f'{nilai:.2f}', va='center', fontsize=10, color='red')

plt.tight_layout()
plt.show()
```

```python
# MELIHAT DATA PAD BANDA ACEH
data_banda_aceh = df[df['Kota'] == 'Kota Banda Aceh']

# Membuat grafik
plt.figure(figsize=(6, 4)) # inchi

# Membuat grafik batang
sns.barplot(x='Tahun', y='PAD', data=data_banda_aceh, palette='colorblind')

plt.title('PAD Kota Banda Aceh per Tahun')
plt.xlabel('Tahun')
plt.ylabel('PAD')

plt.show()
```
